```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr, kendalltau

def format_pvalue(p):
    """p值格式化:极小值时科学计数法(系数4位小数),否则普通4位小数"""
    if p < 0.0001:
        return f"{p:.4e}"
    else:
        return f"{p:.4f}"
# ========= 1) Excel 读列 =========
file_path = r"D:\AllPython\vscodePy\shumo_study\25c\data_clean.xlsx"
sheet_name = "男胎检测数据"
x_col, y_col = "检测孕周_小数周", "Y染色体浓度"
df = pd.read_excel(file_path, sheet_name=sheet_name)
x, y = df[x_col].to_numpy(dtype=float), df[y_col].to_numpy(dtype=float)

# ========= 2) 参数模板 =========
params = {}  # 无超参数

# ========== 2) 计算相关系数 ==========
pearson_res = pearsonr(x, y)
spearman_res = spearmanr(x, y)
kendall_res = kendalltau(x, y)

# 输出结果，相关系数固定4位小数，p值按上述规则
print(f"Pearson:  r = {pearson_res.statistic:.4f},  p = {format_pvalue(pearson_res.pvalue)}")
print(f"Spearman: ρ = {spearman_res.statistic:.4f},  p = {format_pvalue(spearman_res.pvalue)}")
print(f"Kendall:  τ = {kendall_res.statistic:.4f},  p = {format_pvalue(kendall_res.pvalue)}")
```
