```python
import numpy as np
import pandas as pd

# ========= 1) Excel 读列（线性回归示例） =========
file_path = r"your_data.xlsx"
sheet_name = "SGD"
feature_cols = ["x1", "x2"]
target_col = "y"

df = pd.read_excel(file_path, sheet_name=sheet_name)
X = df[feature_cols].to_numpy(dtype=float)
y = df[target_col].to_numpy(dtype=float)

# ========= 2) 参数模板 =========
params = {
    "theta0": np.zeros(X.shape[1]),  # 参数初值
    "lr": 0.01,                      # 学习率
    "n_iter": 1000                   # 迭代次数
}

theta = params["theta0"].copy()
n = len(y)
for _ in range(params["n_iter"]):
    i = np.random.randint(0, n)
    grad_i = (X[i] @ theta - y[i]) * X[i]  # 单样本梯度
    theta -= params["lr"] * grad_i
print("theta=", theta)
```

```python
"""
随机梯度下降

使用方法：
1. 按照下方 TODO 修改 DATA_FILE、列名、参数和输出文件名。
2. 将数据文件放在本脚本同目录，或把 DATA_FILE 改成绝对路径。
3. 运行：python "随机梯度下降.py"
"""

from pathlib import Path
import numpy as np
import pandas as pd



DATA_FILE = "data.csv"  # TODO: 请填写[数据文件路径]，说明：CSV/Excel 均可；若使用 Excel，请在 load_data 中改为 read_excel。
OUTPUT_FILE = "model_output.csv"  # TODO: 请填写[输出文件名]，说明：保存模型结果，建议保留 .csv 或 .xlsx 后缀。
RANDOM_STATE = 42  # TODO: 请填写[随机种子]，说明：用于复现实验；整数即可。
FEATURE_COLUMNS = ["特征1", "特征2"]  # TODO: 请填写[特征列名列表]，说明：数值型。
TARGET_COLUMN = "y"  # TODO: 请填写[目标列名]，说明：连续数值列。
LEARNING_RATE = 0.01  # TODO: 请填写[学习率]，说明：过大可能震荡，过小收敛慢。
EPOCHS = 200  # TODO: 请填写[训练轮数]，说明：正整数。



REQUIRES_DATA = True  # 参数型模型可不提供数据文件；表格型模型必须提供数据。


def load_data() -> pd.DataFrame:
    """读取用户数据；竞赛时通常把 Excel/CSV 表格整理成一行一个样本。"""
    path = Path(DATA_FILE)
    if not path.exists():
        if not REQUIRES_DATA:
            return pd.DataFrame()
        raise FileNotFoundError(
            f"未找到数据文件 {DATA_FILE}。请先修改 DATA_FILE，或将数据放到脚本同目录。"
        )
    if path.suffix.lower() in [".xlsx", ".xls"]:
        return pd.read_excel(path)
    return pd.read_csv(path)


def run_model(data: pd.DataFrame) -> None:
    X = data[FEATURE_COLUMNS].to_numpy(dtype=float)
    y = data[TARGET_COLUMN].to_numpy(dtype=float)
    X = np.column_stack([np.ones(len(X)), X])
    w = np.zeros(X.shape[1])
    rng = np.random.default_rng(RANDOM_STATE)
    for _ in range(EPOCHS):
        for i in rng.permutation(len(X)):
            error = X[i] @ w - y[i]
            w -= LEARNING_RATE * error * X[i]
    print("SGD 参数:", w)


if __name__ == "__main__":
    df = load_data()
    run_model(df)
```
