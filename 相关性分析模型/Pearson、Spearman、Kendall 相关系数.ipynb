```python
import numpy as np
from scipy import stats
import pandas as pd  # 仅矩阵分析时使用，若只做单对变量可注释掉

# ============================================================
# 数据填充区 —— 请将您的数据替换下面的占位符
# ============================================================

# ---------- 单对变量分析 ----------
# 请填入两个变量的观测值，长度需相等，支持列表或numpy数组
# x = [1, 2, 3, 4, 5]          # TODO: 替换为变量X的实际数据
# y = [2, 4, 5, 4, 5]          # TODO: 替换为变量Y的实际数据

# ---------- 多变量矩阵分析（可选） ----------
# 若需计算多个变量之间的相关系数矩阵，请准备一个DataFrame
# 例如从CSV读取：df = pd.read_csv('your_data.csv')
# 下方仅作占位示例，实际使用时请替换为您的数据
df = pd.read_excel(r'D:\AllPython\vscodePy\shumo_study\25c\data_clean.xlsx')   # TODO: 替换
x , y = df[["Y染色体浓度", "检测孕周_小数周"]]

# ============================================================
# 函数定义 —— 无需改动
# ============================================================

def clean_data(x, y, nan_policy='omit'):
    """处理缺失值，返回干净的数组"""
    x = np.array(x, dtype=float)
    y = np.array(y, dtype=float)
    if nan_policy == 'omit':
        mask = ~(np.isnan(x) | np.isnan(y))
        x, y = x[mask], y[mask]
    elif nan_policy == 'raise':
        if np.any(np.isnan(x)) or np.any(np.isnan(y)):
            raise ValueError("数据中存在缺失值，请处理后再试")
    return x, y

def calc_corr(x, y, method='pearson', nan_policy='omit', alpha=0.05):
    """
    计算两个变量之间的相关系数并进行显著性检验
    method: 'pearson' 或 'spearman'
    nan_policy: 'omit' 剔除缺失值, 'raise' 报错
    alpha: 显著性水平，用于标记是否显著
    返回: (相关系数, p值, 是否显著)
    """
    x, y = clean_data(x, y, nan_policy)
    if method == 'pearson':
        corr, p_value = stats.pearsonr(x, y)
    elif method == 'spearman':
        corr, p_value = stats.spearmanr(x, y)
    else:
        raise ValueError("method 必须是 'pearson' 或 'spearman'")
    significant = p_value < alpha
    return corr, p_value, significant

def calc_corr_matrix(data, method='pearson', nan_policy='omit'):
    """
    计算DataFrame各列之间的相关系数矩阵
    method: 'pearson' 或 'spearman'
    返回: 相关系数矩阵(DataFrame)和p值矩阵(DataFrame)
    """
    if method == 'pearson':
        corr_func = lambda a, b: stats.pearsonr(a, b)
    elif method == 'spearman':
        corr_func = lambda a, b: stats.spearmanr(a, b)
    else:
        raise ValueError("method 必须是 'pearson' 或 'spearman'")
    
    cols = data.columns
    n = len(cols)
    corr_mat = pd.DataFrame(np.eye(n), index=cols, columns=cols)
    p_mat = pd.DataFrame(np.zeros((n, n)), index=cols, columns=cols)
    
    for i in range(n):
        for j in range(i+1, n):
            # 提取两列并剔除缺失值
            a = data[cols[i]].values.astype(float)
            b = data[cols[j]].values.astype(float)
            mask = ~(np.isnan(a) | np.isnan(b))
            if nan_policy == 'raise' and mask.sum() < len(a):
                raise ValueError(f"列 {cols[i]} 和 {cols[j]} 存在缺失值")
            a_clean, b_clean = a[mask], b[mask]
            corr, p = corr_func(a_clean, b_clean)
            corr_mat.loc[cols[i], cols[j]] = corr
            corr_mat.loc[cols[j], cols[i]] = corr
            p_mat.loc[cols[i], cols[j]] = p
            p_mat.loc[cols[j], cols[i]] = p
    return corr_mat, p_mat

# ============================================================
# 分析与输出 —— 根据您的需求调用
# ============================================================

print("=" * 50)
print("单对变量相关性分析")
print("=" * 50)

# 皮尔逊相关系数
pearson_r, pearson_p, pearson_sig = calc_corr(x, y, method='pearson')
print(f"皮尔逊相关系数: {pearson_r:.4f}")
print(f"p值: {pearson_p:.4f}  {'* 显著相关' if pearson_sig else '  不显著'}")

# 斯皮尔曼相关系数
spearman_r, spearman_p, spearman_sig = calc_corr(x, y, method='spearman')
print(f"斯皮尔曼相关系数: {spearman_r:.4f}")
print(f"p值: {spearman_p:.4f}  {'* 显著相关' if spearman_sig else '  不显著'}")

# 多变量矩阵分析（如果不需要，可以注释掉以下部分）
# print("\n" + "=" * 50)
# print("多变量相关系数矩阵")
# print("=" * 50)

# pearson_corr_matrix, pearson_p_matrix = calc_corr_matrix(df, method='pearson')
# print("皮尔逊相关系数矩阵:")
# print(pearson_corr_matrix.round(4))
# print("\n对应的p值矩阵:")
# print(pearson_p_matrix.round(4))

# spearman_corr_matrix, spearman_p_matrix = calc_corr_matrix(df, method='spearman')
# print("\n斯皮尔曼相关系数矩阵:")
# print(spearman_corr_matrix.round(4))
# print("\n对应的p值矩阵:")
# print(spearman_p_matrix.round(4))
```
