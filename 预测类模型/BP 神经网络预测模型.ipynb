```python
import numpy as np
import pandas as pd
from sklearn.neural_network import MLPRegressor

# ========= 1) Excel 读列 =========
file_path = r"your_data.xlsx"
sheet_name = "Sheet1"
feature_cols = ["x1", "x2", "x3"]  # list[str]: 特征列名
target_col = "y"                   # str: 目标列名

df = pd.read_excel(file_path, sheet_name=sheet_name)
X_train = df[feature_cols].to_numpy(dtype=float)
y_train = df[target_col].to_numpy(dtype=float)

# 未来样本（你可来自另一张表）
X_future = X_train[-5:, :]  # 这里只是示例

# ========= 2) 参数模板 =========
params = {
    "hidden_layer_sizes": (64, 32),  # tuple[int,...]: 隐藏层结构
    "activation": "relu",            # str: relu/tanh/logistic
    "max_iter": 1000,                # int: 训练迭代上限
    "random_state": 42               # int/None: 随机种子
}

model = MLPRegressor(
    hidden_layer_sizes=params["hidden_layer_sizes"],
    activation=params["activation"],
    solver="adam",
    max_iter=params["max_iter"],
    random_state=params["random_state"]
)
model.fit(X_train, y_train)
y_pred = model.predict(X_future)
print(y_pred)

```

```python
"""
BP 神经网络预测模型

使用方法：
1. 按照下方 TODO 修改 DATA_FILE、列名、参数和输出文件名。
2. 将数据文件放在本脚本同目录，或把 DATA_FILE 改成绝对路径。
3. 运行：python "BP 神经网络预测模型.py"
"""

from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler



DATA_FILE = "data.csv"  # TODO: 请填写[数据文件路径]，说明：CSV/Excel 均可；若使用 Excel，请在 load_data 中改为 read_excel。
OUTPUT_FILE = "model_output.csv"  # TODO: 请填写[输出文件名]，说明：保存模型结果，建议保留 .csv 或 .xlsx 后缀。
RANDOM_STATE = 42  # TODO: 请填写[随机种子]，说明：用于复现实验；整数即可。
TARGET_COLUMN = "y"  # TODO: 请填写[因变量列名]，说明：连续数值列。
TEST_SIZE = 0.2  # TODO: 请填写[测试集比例]，说明：0 到 1。
HIDDEN_LAYERS = (16, 8)  # TODO: 请填写[隐藏层结构]，说明：样本少时不要设太大。
MAX_ITER = 2000  # TODO: 请填写[最大迭代次数]，说明：训练不收敛时可增大。



REQUIRES_DATA = True  # 参数型模型可不提供数据文件；表格型模型必须提供数据。


def load_data() -> pd.DataFrame:
    """读取用户数据；竞赛时通常把 Excel/CSV 表格整理成一行一个样本。"""
    path = Path(DATA_FILE)
    if not path.exists():
        if not REQUIRES_DATA:
            return pd.DataFrame()
        raise FileNotFoundError(
            f"未找到数据文件 {DATA_FILE}。请先修改 DATA_FILE，或将数据放到脚本同目录。"
        )
    if path.suffix.lower() in [".xlsx", ".xls"]:
        return pd.read_excel(path)
    return pd.read_csv(path)


def run_model(data: pd.DataFrame) -> None:
    feature_cols = [c for c in data.columns if c != TARGET_COLUMN]
    X_train, X_test, y_train, y_test = train_test_split(
        data[feature_cols], data[TARGET_COLUMN], test_size=TEST_SIZE, random_state=RANDOM_STATE
    )
    model = Pipeline([
        ("scaler", StandardScaler()),
        ("mlp", MLPRegressor(hidden_layer_sizes=HIDDEN_LAYERS, max_iter=MAX_ITER, random_state=RANDOM_STATE)),
    ])
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    print("RMSE:", mean_squared_error(y_test, pred, squared=False))
    pd.DataFrame({"真实值": y_test, "预测值": pred}).to_csv(OUTPUT_FILE, index=False, encoding="utf-8-sig")


if __name__ == "__main__":
    df = load_data()
    run_model(df)
```
